### 2.1. Раунды финансирования по годам

Задание необходимо выполнить без объединения и дополнительной предобработки на основе датасета `company_and_rounds.csv`.

- Составьте сводную таблицу по годам, в которой на основании столбца `raised_amount` для каждого года указан:
    - типичный размер средств, выделяемый в рамках одного раунда;
    - общее количество раундов финансирования за этот год.
    
- Оставьте в таблице информацию только для тех лет, для которых есть информация о более чем 50 раундах финансирования.
- На основе получившейся таблицы постройте график, который будет отражать динамику типичного размера средств, которые стартапы получали в рамках одного раунда финансирования.

На основе полученных данных ответьте на вопросы:

- В каком году типичный размер собранных в рамках одного раунда средств был максимален?
- Какая тенденция по количеству раундов и выделяемых в рамках каждого раунда средств наблюдалась в 2013 году?

**Составим сводную таблицу по годам, в которой на основании столбца raised_amount для каждого года указан:**

- типичный размер средств, выделяемый в рамках одного раунда (в качестве типичного размера возьмем медиану);
- общее количество раундов финансирования за этот год.

```python
# Составим таблицу: 
df_company_and_rounds_by_years = df_company_and_rounds.copy()
df_company_and_rounds_by_years['year'] = df_company_and_rounds_by_years['funded_at'].dt.year # Извлечем год из даты
df_company_and_rounds_by_years = df_company_and_rounds_by_years.groupby('year')['raised_amount'].agg(['median','count']).reset_index() # Сгруппируем данные по году и рассчитаем значения
df_company_and_rounds_by_years.head()
```

```python
# Оставим информацию только для тех лет, для которых есть информация о более чем 50 раундах финансирования:
df_company_and_rounds_by_years = df_company_and_rounds_by_years.loc[df_company_and_rounds_by_years['count']>50]
df_company_and_rounds_by_years
```

```python
df_company_and_rounds_by_years.info()
```

Построим график, который будет отражать динамику типичного размера средств, которые стартапы получали в рамках одного раунда финансирования.

```python
# Строим график:
import numpy as np
df_company_and_rounds_by_years.plot(kind='line', x='year',y='median',rot=0,legend=False, figsize=(14, 7),
        title=f'Динамика типичного размера средств в рамках однcого раунда финансирования'
)
plt.xlabel('год')
plt.ylabel('типичный размер средств, е6 = млн $')
plt.xticks(np.arange(1999,2013,step=1.0))
plt.grid()
plt.show()
```

Согласно полученный данным можно сделать следующие выводы:

Максимальный типичный размер собранных в рамках одного раунда средств был в 2005 году.

В 2013 наблюдается тенденция увеличения количества раундов и типичного размера выделяемых в рамках каждого раунда средств после продолжительного периода падения этих показателей.


### 3.4. Цены стартапов по категориям

Категории стартапов с типично высокими ценами покупки стартапов и значительным разбросом цен могут быть привлекательными для крупных инвесторов, которые готовы к высоким рискам ради потенциально больших доходов. Среди категорий стартапов выделите категории стартапов, характеризующиеся:

- типично высокими ценами;
- и наибольшим разбросом цен за стартап.

Объясните, почему решили составить топ именно из такого числа категорий и почему рассчитывали именно так.

В качестве стартапов с типично высокими ценами выберем стартапы с суммой финансирования в диапазоне от 75% процентиля до верхней границы выбросов. Значения выше будут выбросами и будут искажать данные.

Выведем топ категорий стартапов с типично высокими ценами и наибольшим разбросом цен за стартап. Выберем топ-5 категорий, так как этого достаточно, чтобы выделить основные категории.

В качестве рабочей таблицы возьмем таблицу без компаний с нулевой суммой финансирования - df_slice_companies_unique_without_null_funding.

```python
# Сформируем срез компаний с суммой финансирования в диапазоне от 75% процентиля (Q3) до верхней границы выбросов (upper_bound):
top = df_slice_companies_unique_without_null_funding.loc[(df_slice_companies_unique_without_null_funding['funding_total']>Q3) & (df_slice_companies_unique_without_null_funding['funding_total']<upper_bound)].groupby('category_code')['funding_total'].agg({'median','std'}).reset_index()
# Посчитаем медиану и разброс для значений суммы финансирования:
top = top.rename(columns={'median': 'median_funding', 'std': 'funding_std_dev'})
top = top.sort_values('funding_std_dev', ascending=False)
top.head()
```

```python
# Визуализируем топ-5:
top.plot(kind='barh', 
         y='funding_std_dev', 
         x='category_code',
        legend=False)
plt.xlabel('сумма финансирования, 10 млн $')
plt.ylabel('категория компании')
plt.title('Топ-5 категорий компаний с типично высокими ценами и наибольшим разбросом цен за стартап')
plt.ylim(4.5,-0.5)
figsize=(7,5)
plt.show()
```
